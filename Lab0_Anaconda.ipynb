```python
scores = [85, 92, 78, 90, 88]   # pretend these are 5 quiz scores
print("Scores:", scores)
```

```python
print("Average:", sum(scores) / len(scores))
print("Minimum:", min(scores))
print("Maximum:", max(scores))
```

```python
import matplotlib.pyplot as plt
 
quizzes = ["Quiz 1", "Quiz 2", "Quiz 3", "Quiz 4", "Quiz 5"]
 
plt.figure(figsize=(6, 4))
plt.bar(quizzes, scores, color="steelblue")
plt.xlabel("Quiz")
plt.ylabel("Score")
plt.title("My Quiz Scores")
plt.savefig("my_first_chart.png", dpi=150, bbox_inches="tight")
plt.show()
```
